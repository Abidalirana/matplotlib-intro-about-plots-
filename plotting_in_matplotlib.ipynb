# 1. LINE PLOT

```python
import matplotlib.pyplot as plt      #import liberary
x=[1,2,3,4,5,6,7]                # both matrax elments should be a same
y=[12,34,67,12,45,12,45]
```

```python
fig,sub_plot=plt.subplots()
sub_plot.plot(x,y);
```

# 2 Scatter plots

```python
fig,sub_plot=plt.subplots()
sub_plot.scatter(x,y)
```

# 3 Bar plots 
### for showing items values | named parameter values tell bar plot
### format should be item and values 

```python
shop={'item1':20,'item2':30,'item3':10}             #vertical bar plot
fig,sub_plot=plt.subplots()
sub_plot.bar(shop.keys(),shop.values())
     
```

### if we want to make horizontal bar plot 

```python
shop={'item1':20,'item2':30,'item3':10,}     # for horizintal bar plots
fig,sub_plot=plt.subplots()
sub_plot.barh(list(shop.keys()),list(shop.values())) 
plt.show()
```

# 4 Histogrm
### tell about data distributions | Distribution of random sample

```python
fig,sub_plot=plt.subplots()
sub_plot.hist(x);
```
